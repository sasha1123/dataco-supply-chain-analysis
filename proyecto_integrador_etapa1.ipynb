## 3. Identificación de valores faltantes

Se revisa la presencia de valores faltantes en cada variable del conjunto de datos para identificar aquellas columnas que requieren algún tratamiento antes del análisis.

```python
valores_faltantes = df.isnull().sum().sort_values(ascending=False)

valores_faltantes
```

```python
valores_faltantes[valores_faltantes > 0]
```

```python
resumen_faltantes = pd.DataFrame({
    "Valores faltantes": df.isnull().sum(),
    "Porcentaje (%)": (df.isnull().mean() * 100).round(4)
})

resumen_faltantes = resumen_faltantes[
    resumen_faltantes["Valores faltantes"] > 0
].sort_values("Porcentaje (%)", ascending=False)

resumen_faltantes
```

## 5. Revisión estadística inicial

Se realiza una revisión descriptiva de las variables numéricas para conocer sus principales características, como valores mínimos, máximos, promedio y dispersión antes de realizar las transformaciones de los datos.

## 6. Limpieza de valores faltantes

Antes de realizar modificaciones se crea una copia del conjunto de datos original. Posteriormente se tratan las variables con una proporción elevada de valores faltantes, conservando el archivo original sin modificaciones.

```python
df_limpio = df.copy()

df_limpio.shape
```

```python
df_limpio = df_limpio.drop(
    columns=["Product Description", "Order Zipcode"]
)

df_limpio.shape
```

```python
df_limpio.isnull().sum()[
    df_limpio.isnull().sum() > 0
]
```

## 7. Transformación de variables de fecha

Se revisan las variables relacionadas con fechas para verificar su tipo de dato y transformarlas a un formato temporal adecuado para el análisis.

```python
df_limpio[
    ["order date (DateOrders)", "shipping date (DateOrders)"]
].head()
```

```python
df_limpio[
    ["order date (DateOrders)", "shipping date (DateOrders)"]
].dtypes
```

```python
df_limpio["order date (DateOrders)"] = pd.to_datetime(
    df_limpio["order date (DateOrders)"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

df_limpio["shipping date (DateOrders)"] = pd.to_datetime(
    df_limpio["shipping date (DateOrders)"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)
```

```python
df_limpio[
    ["order date (DateOrders)", "shipping date (DateOrders)"]
].dtypes
```

```python
df_limpio[
    ["order date (DateOrders)", "shipping date (DateOrders)"]
].isnull().sum()
```

```python
df_limpio.columns.tolist()
```

## 8. Tratamiento de valores faltantes restantes

Después de la limpieza inicial permanecen valores faltantes en las variables Customer Lname y Customer Zipcode. Debido a que estas variables no son necesarias para el objetivo principal del análisis, se eliminan del conjunto de datos en lugar de completar la información con valores que no pueden determinarse a partir de los datos disponibles.

```python
df_limpio = df_limpio.drop(
    columns=["Customer Lname", "Customer Zipcode"]
)

df_limpio.shape
```

```python
df_limpio.isnull().sum().sum()
```

```python
print("Registros:", df_limpio.shape[0])
print("Columnas:", df_limpio.shape[1])
print("Valores faltantes:", df_limpio.isnull().sum().sum())
```

## 9. Estandarización de los nombres de las variables

Se estandarizan los nombres de las columnas para facilitar su utilización durante el análisis. Se eliminan espacios innecesarios, se utilizan letras minúsculas y los espacios o caracteres especiales se sustituyen por guiones bajos.

```python
columnas_antes = df_limpio.columns.tolist()

df_limpio.columns = (
    df_limpio.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[()]", "", regex=True)
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)
```

```python
comparacion_columnas = pd.DataFrame({
    "Nombre original": columnas_antes,
    "Nombre estandarizado": df_limpio.columns
})

comparacion_columnas.head(15)
```

## 10. Verificación final del conjunto de datos

Se realiza una revisión final de la estructura del conjunto de datos después de los procesos de limpieza y transformación, con el propósito de comprobar que no existan valores faltantes y que los tipos de datos y nombres de variables sean adecuados para continuar con el análisis.

```python
print("Registros finales:", df_limpio.shape[0])
print("Columnas finales:", df_limpio.shape[1])
print("Valores faltantes:", df_limpio.isnull().sum().sum())
print("Registros duplicados:", df_limpio.duplicated().sum())
```

## 11. Exportación del conjunto de datos preparado

Una vez finalizados los procesos de limpieza y transformación, se guarda una nueva versión del conjunto de datos en formato CSV. El archivo original se conserva sin modificaciones y la nueva versión será utilizada para continuar con la segunda etapa del Proyecto Integrador.

```python
df_limpio.to_csv(
    "DataCoSupplyChainDataset_limpio.csv",
    index=False,
    encoding="utf-8"
)
```
